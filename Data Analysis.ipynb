# ***Progress Bars***

```python
import time
from tqdm import tqdm

for i in tqdm(range(0,100)):
  time.sleep(0.01)
```

# ***Statistics***

**Statistics with Python**

Find Mean

```python
import numpy
values = [4, 11, 7, 14]
x = numpy.mean(values)
print(x)
```

Find Mode

```python
from statistics import multimode

values = [4,7,3,8,11,7,10,19,6,9,12,12]

x = multimode(values)

print(x)
```

```python
from statistics import multimode

name = ['Alice','John','Bob','Maria','John','Julia','Carol']

y = multimode(name)

print(y)
```

Find Median

```python
import numpy

values = [13,21,21,40,42,48,55,72]

x = numpy.median(values)

print(x)
```
